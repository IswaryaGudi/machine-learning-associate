```python
# Creating A Series in Pandas
# to create a series we use the pd.Series() function.
```

```python
# creating a series from a list -- A list is a mutable collection of elements
l = [45,78,56,445,78]
s = pd.Series(l)
s    
```

```python
# creating a series from a tuple -- An tuple is an immutable ordered sequence of elements
t = (10,13,56,88)
print(pd.Series(t))
```

```python
# creating a series from a numpy array -- A numpy array is a fast and efficient way to store numerical data.
import numpy as np
arr=np.array([1,3,5,7,9])
print(pd.Series(arr))
```

```python
# Creating a Series from a dictionary -- A dictionary has a key-value pairs where keys acts as index labels in the series
d = {'a':1,'b':2,'c':3,'d':4}
print(pd.Series(d))
```

```python
# create a series of sets and transform/convert that to pandas
set1 = {1,5,6,7,9,100}
list1 = list(set1)
print(pd.Series(list1))
```
